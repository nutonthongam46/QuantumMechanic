```python

import numpy as np
import matplotlib.pyplot as plt


m = 940
h = 197.3
k = 100
C = h**2 / (2*m)


N = 500
ri = 0.001
rf = 20

dr = (rf-ri)/(N+1)
r = ri + dr*np.arange(1, N+1)


b_values = [0, 10, 20]

plt.figure(figsize=(8,5))

for b in b_values:

    
    V = (k*r**2)/2 + (b*r**3)/2

    
    main = 2*C/dr**2 + V
    side = -C/dr**2

    H = np.diag(main)
    H += np.diag(np.full(N-1, side), 1)
    H += np.diag(np.full(N-1, side), -1)

    
    E, psi = np.linalg.eigh(H)

    
    u = psi[:, 0]

    
    u = u / np.sqrt(np.sum(u**2)*dr)

    
    print("b =", b, "Energy =", E[0])

    
    plt.plot(r, u, label="b = "+str(b))

plt.xlabel("Radial distance r")
plt.ylabel("Wavefunction u(r)")
plt.title("Harmonic Oscillator")
plt.legend()
plt.grid()
plt.show()
```
