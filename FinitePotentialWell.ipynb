```python

import numpy as np
import matplotlib.pyplot as plt


h = 1.05457e-34
me = 9.10938e-31
ev = 1.60218e-19


mass = 0.067 * me
V = 0.3 * ev
a = 4e-9


z0 = a * np.sqrt(2 * mass * V) / h



def even(z):
    return z*np.tan(z) - np.sqrt(z0**2-z**2)

def odd(z):
    return -z/np.tan(z) - np.sqrt(z0**2-z**2)



def bisection(f, x1, x2):
    for i in range(100):
        mid = (x1+x2)/2

        if f(x1)*f(mid) < 0:
            x2 = mid
        else:
            x1 = mid

    return (x1+x2)/2



eps = 1e-8

roots = [
    (bisection(even, eps, np.pi/2-eps), "Even"),
    (bisection(odd, np.pi/2+eps, z0-eps), "Odd")
]


x = np.linspace(-3*a, 3*a, 1000)

for z, parity in roots:

    energy = -V + h**2*z**2/(2*mass*a**2)
    energy = energy/ev

    k = z/a
    kap = np.sqrt(-2*mass*energy*ev)/h

    if parity == "Even":
        A = 1/np.sqrt(a + a*np.sin(2*z)/(2*z)
                      + np.cos(z)**2/kap)

        psi = np.where(
            abs(x) <= a,
            A*np.cos(k*x),
            A*np.cos(z)*np.exp(-kap*(abs(x)-a))
        )

    else:
        A = 1/np.sqrt(a - a*np.sin(2*z)/(2*z)
                      + np.sin(z)**2/kap)

        psi = np.where(
            abs(x) <= a,
            A*np.sin(k*x),
            np.sign(x)*A*np.sin(z)*np.exp(-kap*(abs(x)-a))
        )

    print(parity, "Energy =", round(energy, 5), "eV")

    plt.plot(x*1e9, psi, label=parity)

plt.axvline(-a*1e9, linestyle="--")
plt.axvline(a*1e9, linestyle="--")
plt.xlabel("Position (nm)")
plt.ylabel("Wavefunction")
plt.title("Finite Square Potential Well - GaAs")
plt.legend()
plt.grid()
plt.show()



E = np.linspace(1.01*V, 3*V, 500)

q = np.sqrt(2*mass*(E-V))/h

T = 1/(1 + (V**2/(4*E*(E-V)))*np.sin(2*a*q)**2)

plt.plot(E/ev, T)
plt.xlabel("Energy (eV)")
plt.ylabel("Transmission Coefficient")
plt.title("Transmission through Finite Potential Barrier")
plt.grid()
plt.show()
```
