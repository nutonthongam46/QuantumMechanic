```python
import numpy as np 



def matrix_1(n):
    H=np.zeros((n,n))


    for i in range(n):
        H[i,i]= -2

    for j in range(n-1):
        H[j,j+1]= 1
        H[j+1,j]=1

    print(H)

matrix_1(500)







```
